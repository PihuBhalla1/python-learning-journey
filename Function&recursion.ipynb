```python
def calc_sum(a,b):
    sum=a+b
    print(sum)
    return(sum)
calc_sum(2, 3)
```

```python
def calc_sum(a,b): #function defination(define) nd a,b are parameters
    return(a+b)
sum=calc_sum(9,2)   #function call;9nd2 are arguments
print(sum)
```

```python
#average of 3 numbers
def avg(a,b,c):
    sum=a+b+c
    avg=sum/3
    print(avg)
    return avg
avg(99,98,90)
```

```python
def cal_prod(a=2,b=5):
    print(a*b)
    return a*b
cal_prod()
```

```python
def cal_prod(a,b=7):
    print(a*b)
    return a*b
cal_prod(2)
```

```python
def cal_prod(b,a=5):   #if we do a=7,b this will give error
    print(a*b)
    return a*b
cal_prod(7)
```

```python
#WAP to print the length of a list(list is param)
city=["delhi","pune","mumbai","chennai","noida"]
def len_list(list):
    print(len(list))
len_list(city)
```

```python
#WAF to print the elements of a list in a single line(list is the para)
city=["delhi","pune","mumbai","chennai","noida"]
def print_list(list):
    for item in list:
        print(item,end=" ")
print_list(city)
```

```python
#WAP to find the factorial of n(n is param)
def cal_fact(n):
    fact=1
    for i in range(1,n+1):
        fact *=i
    print(fact)
cal_fact(5)
```
