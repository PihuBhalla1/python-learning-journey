## If Else

```python
a=int(input("Enter your age:"))
if(a>=18):
    print("Adult")
    print("Good for you")
else:
    print("Not Adult")
print("End Of Program")
```

## If Elif Else Ladder

```python
a=int(input("Enter your age:"))
if(a>=18):
    print("Adult")
    print("Good for you")
elif(a<0):
    print("You are entering invalid negative age")
elif(a==0):
    print("You are entering 0 which is not a valid age")
else:
    print("Not Adult")
print("End Of Program")
```

Q write a python program to find the greatest of four numbers entered by user

```python
a=int(input("Enter Number 1:"))
b=int(input("Enter Number 2:"))
c=int(input("Enter Number 3:"))
d=int(input("Enter Number 4:"))
if(a>b and a>c and a>d):
    print("Greatest number is a:", a)
if(b>a and b>c and b>d):
    print("Greatest number is b:", b)
if(c>b and c>a and c>d):
    print("Greatest number is c:", c)
if(d>b and d>c and d>a):
    print("Greatest number is d:", d)
```
