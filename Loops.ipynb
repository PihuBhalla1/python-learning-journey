## FOR LOOP

IT RUNS ITEM BY ITEM AND MARK ITSELF COMPLETE ONCE ALL ITEMS EXECUTED SUCCESFULLY

Iteration  
Iteration ka matlab hai kisi collection ke har item ko ek-ek karke process karna.

EG:[V1,V2,V3,V4.....]

ITERABLE(common iterables):STR,LIST,TUPLE,SET,DICTIONARY... (Jis object ke items ko ek-ek karke access kiya ja sakta hai use iterable kehte hai)

ITERATE VALUE:ITEM   (Loop ke har round mein jo current item milta hai usse iterate value kehte hain)

```python
X=[37,52,88,99]    #list is iterable and each value is iterate value
for i in X:        #i is a name of iterate value inside iterable
    print(i/2)     #this code will execute with all items in a list
#hence for loop works item by item and once all items are completed then it will stops
```

### Nested for Loop

```python
for i in range(1,5): #outer
    for j in range(1,3):  #nested
        print(j)

#1)outer loop will run iteration
#2))inner loop will run completely
#3)outer will run next iteration
#if outer will run comppletely
#this will followed again and again 
```

```python
for i in range(1,10,2): #outer
    for j in range(1,3): #Nested
        print(j,i)

#1)outer loop will run iteration
#2)inner loop will run completely
#3)outer will run next iteration
#4)inner loop will run completely
#this will followed again nd again till th time
        #akk items of outer loop executed
#if outer loop is having 4 values
#then inner loop will run 4 times completely
```

# Enumerate

```python
products=['Laptop','Mouse','Monitor','Keyboard']
price=[35000,500,1200,800]
qty=[10,8,7,5]
```

```python
list(enumerate(products))
```

```python
index=3
for i,v in enumerate(products):
    if i==index:
        print(v)
```

```python
start_index=1
end_index=3
for i,v in enumerate(products):
    if i in range(start_index,end_index):
        print(v)
```

# Zip

```python
list(zip(products,price,qty)) #item by item of all lists will be bagged in tuple
```

```python
for P,pr,Q in zip(products,price,qty):
    #taking items of all lists where P is product,pr is price,Q is qty
    print(f'Product:{P}||Price:{pr}||Qty:{Q} || Sale:{pr*Q}')
#printing in format string text written outside curly brackets will not considered as code
```

```python
total_sales=0   #initially total_sales is xero
for P,Q in zip(price,qty):   #in loop P means price_item,Q means qty_item
    total_sales+=P*Q   #sales=Price X Qty
```

```python
total_sales
```

# Pass|Continue

```python
X=[1,57,9,87,32,897,1,79,13897,1,978,13,98,13]
```

```python
for i in X:
    if i==1:
            pass
    else:
        print(i)
```

```python
for i in X:
    if i==1:
            pass  #skip the line of code
            print(i)
    else:
        print(i)
```

```python
for i in X:
    if i==1:
        continue #skip the block of code
        print(i)
    else:
        print(i)
```
